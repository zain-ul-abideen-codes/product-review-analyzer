## Text Cleaning
start NLP Preprocessing

```python
# Import regular expressions for text cleaning
import re


# Function to clean review text
def clean_text(text):

    # Convert text to lowercase
    text = text.lower()

    # Remove punctuation and special characters
    text = re.sub(r"[^a-zA-Z0-9\s]", "", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
# Apply the cleaning function to all reviews
df["cleaned_review"] = df["reviewText"].apply(clean_text)

# Display original and cleaned reviews
df[["reviewText", "cleaned_review"]].head(10)
```

## Step 16.2 — Create Initial Topic Labels

Har review ke text ko check kiya jayega:

- Agar Price ke keywords milen → Price
- Agar Delivery ke keywords milen → Delivery
- Agar Customer Service ke keywords milen → Customer Service

In labels ko baad mein topic classification model train karne ke liye use kiya jayega.

```python
# Function to assign a topic to a review
def assign_topic(text):
    # Check each topic and its keywords
    for topic, keywords in topic_keywords.items():
        
        # Check whether any keyword exists in the review
        for keyword in keywords:
            if keyword in text:
                return topic
    
    # If no topic keyword is found
    return "Other"


# Apply the topic labeling function to all cleaned reviews
df["topic"] = df["cleaned_review"].apply(assign_topic)

# Display reviews with their assigned topics
df[["cleaned_review", "topic"]].head(10)
```

## Step 16.7 — Improve Topic Assignment

Previous method mein review ka sirf pehla matching keyword check hota tha.

Ab hum:
1. Har topic ke matching keywords count karenge.
2. Jis topic ke sabse zyada matches honge, woh topic assign hoga.
3. Agar koi keyword match nahi hoga, to `Other` assign hoga.

Is approach se multiple-topic reviews ko better handle kiya ja sakega.

```python
# Function to assign a topic based on the number of matching keywords
def assign_topic(text):
    
    # Store the number of matches for each topic
    topic_scores = {}

    # Check every topic
    for topic, keywords in topic_keywords.items():
        
        # Count matching keywords
        score = 0
        
        for keyword in keywords:
            if keyword in text:
                score += 1
        
        # Save the score for this topic
        topic_scores[topic] = score

    # Find the topic with the highest score
    best_topic = max(topic_scores, key=topic_scores.get)

    # If no keyword matched, assign Other
    if topic_scores[best_topic] == 0:
        return "Other"

    return best_topic


# Apply the improved topic assignment
df["topic"] = df["cleaned_review"].apply(assign_topic)

# Count reviews in each topic
topic_counts = df["topic"].value_counts()

# Display the new topic distribution
print("New Topic Distribution:")
print(topic_counts)
```

## Phase 3 — Topic Classification

In this phase, we will train a machine learning model to automatically classify
each product review into a topic.

The topics are:
- Packaging
- Delivery
- Customer Service
- Price
- Performance / Compatibility
- Quality
- Other

We will use TF-IDF to convert review text into numerical features and
Logistic Regression for topic classification.

```python
from sklearn.model_selection import train_test_split

# Review text will be the input
X_topic_text = df["cleaned_review"]

# Topic labels will be the target
y_topic = df["topic"]

# Split the data into training and testing sets
X_topic_train_text, X_topic_test_text, y_topic_train, y_topic_test = train_test_split(
    X_topic_text,
    y_topic,
    test_size=0.2,
    random_state=42,
    stratify=y_topic
)

# Display the number of samples
print("Training samples:", len(X_topic_train_text))
print("Testing samples:", len(X_topic_test_text))
```

### Step 3.3 — Convert Review Text into TF-IDF Features

Machine learning models cannot directly understand text.

TF-IDF converts each review into numerical features based on the importance
of words in the review dataset.

We will:
- fit TF-IDF on training reviews
- transform training reviews
- transform testing reviews

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Create the TF-IDF vectorizer
topic_vectorizer = TfidfVectorizer()

# Learn vocabulary and convert training reviews into TF-IDF features
X_topic_train = topic_vectorizer.fit_transform(X_topic_train_text)

# Convert testing reviews using the same vocabulary
X_topic_test = topic_vectorizer.transform(X_topic_test_text)

# Display the shapes of the feature matrices
print("Training TF-IDF shape:", X_topic_train.shape)
print("Testing TF-IDF shape:", X_topic_test.shape)
```

### Step 3.4 — Train the Topic Classification Model

We will use Logistic Regression to learn the relationship between
review text features and their topic labels.

The model will learn patterns from the TF-IDF features and predict
one of our predefined topics for a new review.

```python
from sklearn.linear_model import LogisticRegression

# Create the Logistic Regression topic classification model
topic_model = LogisticRegression(max_iter=1000)

# Train the model using TF-IDF features and topic labels
topic_model.fit(X_topic_train, y_topic_train)

print("Topic classification model trained successfully.")
```

### Step 3.5 — Make Topic Predictions

The trained model will now predict topics for the unseen test reviews.

The model will predict one of the predefined topics for each review.

```python
# Predict topics for the test reviews
y_topic_pred = topic_model.predict(X_topic_test)

# Display the first 10 predicted topics
print("Predicted topics:")
print(y_topic_pred[:10])
```

### Step 3.6 — Evaluate the Topic Classification Model

We will evaluate the topic classification model using:

- Accuracy
- Precision
- Recall
- F1-score
- Macro F1-score

These metrics help us understand how well the model predicts each topic.

```python
# Import evaluation metrics
from sklearn.metrics import accuracy_score, classification_report

# Calculate topic classification accuracy
topic_accuracy = accuracy_score(y_topic_test, y_topic_pred)

# Display accuracy
print("Topic Accuracy:", topic_accuracy)

# Display precision, recall, and F1-score for each topic
print("\nTopic Classification Report:")
print(classification_report(y_topic_test, y_topic_pred))
```

### Step 3.7 — Handle Class Imbalance

The topic dataset contains an unequal number of reviews in each topic.

To give smaller classes more importance, we will use
`class_weight="balanced"` in Logistic Regression.

This will help the model pay more attention to underrepresented topics.

```python
from sklearn.linear_model import LogisticRegression

# Create a new Logistic Regression model
# class_weight="balanced" gives more importance to smaller classes
topic_model_balanced = LogisticRegression(
    max_iter=1000,
    class_weight="balanced"
)

# Train the balanced model
topic_model_balanced.fit(
    X_topic_train,
    y_topic_train
)

print("Balanced topic classification model trained successfully.")
```

### Step 3.8 — Make Predictions with the Balanced Model

The balanced topic classification model will predict topics
for the unseen test reviews.

We will later compare these predictions with the previous model.

```python
# Predict topics using the balanced model
y_topic_pred_balanced = topic_model_balanced.predict(X_topic_test)

# Display the first 10 predictions
print("Balanced model predictions:")
print(y_topic_pred_balanced[:10])
```

### Step 3.9 — Evaluate the Balanced Topic Model

We will evaluate the balanced topic classification model using:

- Accuracy
- Precision
- Recall
- F1-score
- Macro F1-score

We will compare these results with the original topic model.
